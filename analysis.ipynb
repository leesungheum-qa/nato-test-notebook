# 성적 분석 노트북
학생 점수의 평균과 최고 점수를 구합니다.

```python
scores = [72, 85, 90, 64, 98]
names = ['민수', '서연', '지훈', '유진', '하늘']
```

```python
def calc_average(values):
    return sum(values) / len(values)

print(calc_average(scores))
```

```python
best = max(scores)
best_name = names[scores.index(best)]
print(best_name, best)
```
